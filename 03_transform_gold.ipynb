Ten skrypt oczyszcza i transformuje dane w hurtowni

```python
%sql
UPDATE dw.dim_car
SET fuel_type = 
  CASE LEFT(LOWER(fuel_type), 1)
    WHEN 'd' THEN 'Diesel'
    WHEN 'p' THEN 'Petrol'
    ELSE NULL END
WHERE fuel_type not in ('Diesel', 'Petrol');
```
