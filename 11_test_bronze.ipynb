This notebook runs the **bronze** data quality tests (`tests/test_bronze.py`) with pytest.
Run it after `10_extract_bronze` and before `20_transform_silver`.

Checks that the extract is complete: every landing table exists and is not empty, has the columns the pipeline uses, matches the MySQL row counts and has unique primary keys. Bad source values are only reported as warnings - bronze keeps the raw data as it is.

* checks with severity `error` fail the test - the notebook raises an exception, which stops the job
* checks with severity `warning` pass, but are listed in the pytest warnings summary
* only blocking checks: add `"-m", "critical"` to the pytest arguments; a single check: `"-k", "C01"`

```python
import sys
import pytest

sys.dont_write_bytecode = True  # workspace files are read-only, so pytest must not write __pycache__

exit_code = pytest.main(["tests/test_bronze.py", "-v", "--tb=short", "-rfE", "-p", "no:cacheprovider"])
if exit_code != pytest.ExitCode.OK:
    raise Exception(f"Bronze data quality tests failed (pytest exit code {exit_code})")
```
