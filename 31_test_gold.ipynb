This notebook runs the **gold** data quality tests (`tests/test_gold.py`) with pytest.
Run it after `30_load_gold` and before `reports`.

Checks the dimensional model (surrogate and foreign keys, SCD2 history, computed measures, date dimensions, bridge tables) and reconciles it end to end with the source.

* checks with severity `error` fail the test - the notebook raises an exception, which stops the job
* checks with severity `warning` pass, but are listed in the pytest warnings summary
* only blocking checks: add `"-m", "critical"` to the pytest arguments; a single check: `"-k", "C01"`

```python
import sys
import pytest

sys.dont_write_bytecode = True  # workspace files are read-only, so pytest must not write __pycache__

exit_code = pytest.main(["tests/test_gold.py", "-v", "--tb=short", "-rfE", "-p", "no:cacheprovider"])
if exit_code != pytest.ExitCode.OK:
    raise Exception(f"Gold data quality tests failed (pytest exit code {exit_code})")
```
