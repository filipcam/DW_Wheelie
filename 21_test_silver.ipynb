This notebook runs the **silver** data quality tests (`tests/test_silver.py`) with pytest.
Run it after `20_transform_silver` and before `30_load_gold`.

Main quality gate before the gold load: checks that the cleansing rules were applied and that staging is internally consistent (unique business keys, valid lookups, equipment groups).

* checks with severity `error` fail the test - the notebook raises an exception, which stops the job
* checks with severity `warning` pass, but are listed in the pytest warnings summary
* only blocking checks: add `"-m", "critical"` to the pytest arguments; a single check: `"-k", "C01"`

```python
import sys
import pytest

sys.dont_write_bytecode = True  # workspace files are read-only, so pytest must not write __pycache__

exit_code = pytest.main(["tests/test_silver.py", "-v", "--tb=short", "-rfE", "-p", "no:cacheprovider"])
if exit_code != pytest.ExitCode.OK:
    raise Exception(f"Silver data quality tests failed (pytest exit code {exit_code})")
```
