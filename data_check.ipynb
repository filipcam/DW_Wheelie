```python
%sql
SELECT c.car_production_year,Year(d.rental_date),d.rental_date,f.*
FROM dw.fact_rental f
INNER JOIN dw.dim_car c ON c.car_key=f.car_key
INNER JOIN dw.dim_rental_date d ON d.rental_date_key=f.rental_date_key  
WHERE Year(d.rental_date)< c.car_production_year
```
